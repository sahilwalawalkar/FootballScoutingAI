```python
passes = pd.read_csv(
    os.path.join(data_path, "player_accurate_passes.csv")
)

passes.head()
```

```python
print("Dataset shape:", passes.shape)

print("\nColumns:")
print(passes.columns.tolist())

print("\nData types:")
print(passes.dtypes)
```

```python
master_df = passes.drop(columns=["Rank"]).copy()

master_df.head()
```

```python
master_df.shape
```

```python
goals = pd.read_csv(
    os.path.join(data_path, "player_top_scorers.csv")
)

goals.head()
```

```python
goals = goals[
    ["Player", "Team", "Goals", "Penalties"]
]

goals.head()
```

```python
master_df = master_df.merge(
    goals,
    on=["Player", "Team"],
    how="outer"
)

master_df.head()
```

```python
how="outer"
```

```python
xg = pd.read_csv(
    os.path.join(data_path, "player_expected_goals.csv")
)

xg.head()
```

```python
xg = xg[
    ["Player", "Team", "Expected Goals (xG)"]
]

master_df = master_df.merge(
    xg,
    on=["Player", "Team"],
    how="outer"
)

master_df.head()
```

```python
ratings = pd.read_csv(
    os.path.join(data_path, "player_player_ratings.csv")
)

ratings.head()
```

```python
ratings = ratings[
    [
        "Player",
        "Team",
        "FotMob Rating",
        "Player of the Match Awards"
    ]
]

master_df = master_df.merge(
    ratings,
    on=["Player", "Team"],
    how="outer"
)

master_df.head()
```
